```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv('wbc.csv')
```

```python
df_cleaned = df.drop(columns=['id', 'Unnamed: 32'])
```

```python
le = LabelEncoder()
df_cleaned['diagnosis'] = le.fit_transform(df_cleaned['diagnosis'])
```

```python
kolom_numerik = df_cleaned.columns.drop('diagnosis')
```

```python
scaler = StandardScaler()
df_cleaned[kolom_numerik] = scaler.fit_transform(df_cleaned[kolom_numerik])
```
