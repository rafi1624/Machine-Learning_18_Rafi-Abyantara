```python
import zipfile
import os

zip_path = "/content/dataset.zip"
extract_path = "/content/dataset_extracted"

if os.path.exists(zip_path):
    try:
        print("Mengekstrak file... Silakan tunggu.")
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_path)
        print(f"Ekstraksi selesai! File berhasil diekstrak ke: {extract_path}")
        print("Isi folder hasil ekstraksi:")
        print(os.listdir(extract_path)[:10]) # Tampilkan 10 item pertama
    except zipfile.BadZipFile:
        print("Error: File ZIP masih rusak atau tidak lengkap (BadZipFile).")
        print("Solusi: Pastikan Anda telah menghapus file lama, mengupload ulang, dan menunggu indikator lingkaran upload di pojok kiri bawah selesai sepenuhnya.")
else:
    print(f"File {zip_path} tidak ditemukan. Silakan upload file Anda terlebih dahulu.")
```

```python
import tensorflow as tf
# Mengimpor ImageDataGenerator dari modul legacy keras
from keras.src.legacy.preprocessing.image import ImageDataGenerator
```

```python
	train_datagen = ImageDataGenerator(rescale = 1./255,
	                                   shear_range = 0.2,
	                                   zoom_range = 0.2,
	                                   horizontal_flip = True)
	training_set = train_datagen.flow_from_directory('dataset_extracted/dataset/training_set',
	                                                 target_size = (64, 64),
	                                                 batch_size = 32,
	                                                 class_mode = 'binary')
```

```python
	test_datagen = ImageDataGenerator(rescale = 1./255)
	test_set = test_datagen.flow_from_directory('dataset_extracted/dataset/test_set',
	                                            target_size = (64, 64),
	                                            batch_size = 32,
	                                            class_mode = 'binary')
```

```python
cnn = tf.keras.models.Sequential()
```

```python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu', input_shape=[64, 64, 3]))
```

```python
	cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu'))
	cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```

```python
cnn.add(tf.keras.layers.Flatten())
```

```python
cnn.add(tf.keras.layers.Dense(units=128, activation='relu'))
```

```python
cnn.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

```python
cnn.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```
